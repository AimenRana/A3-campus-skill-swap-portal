The import jason code is used 

```python
import json
import os

# The name of the file where student data will be saved
FILENAME = "students_data.json"

```

## Load Data Function 

```python
# 1. Load data function
def load_data():
  if not os.path.exists(FILENAME): 
    save_data(INITIAL_STUDENTS_DATA)
    return INITIAL_STUDENTS_DATA

  try:
    with open(FILENAME, "r") as file:
      data = json.load(file)
      if not data or not isinstance(data, dict):
        save_data(INITIAL_STUDENTS_DATA)
        return INITIAL_STUDENTS_DATA
      return data
  except Exception:
    save_data(INITIAL_STUDENTS_DATA)
    return INITIAL_STUDENTS_DATA
```

## Data Saving Function

```python
# 2. Save data function
def save_data(data):
  with open(FILENAME, "w") as file:
    json.dump(data, file, indent=4)
```

## Registeration for New Users

```python
# 3. Register a new user
def register():
  print("\n==============================")
  print("   REGISTRATION PAGE          ")
  print("==============================")
  data = load_data()

  user_id = input("Enter a new User ID: ").strip().lower()

  if not user_id:
    print("Error: User ID cannot be empty.")
    return

  if user_id in data:
    print("Error: This User ID already exists! Please try logging in.")
    return

  password = input("Enter a password: ")

  skills_input = input(
      "Enter your skills separated by commas (e.g., Python, Math, Design): "
  )
  skills = [s.strip() for s in skills_input.split(",") if s.strip()]

  data[user_id] = {"password": password, "skills": skills}

  save_data(data)
  print("Registration successful! You can now log in.")
```

## Log in for Existing User

```python
# 4. Log in an existing user
def login():
  print("\n==============================")
  print("   LOGIN PAGE                 ")
  print("==============================")
  data = load_data()

  user_id = input("Enter your User ID: ").strip().lower()
  password = input("Enter your password: ")

  if user_id in data and data[user_id]["password"] == password:
    print(f"\n[+] Login Successful! Welcome, {user_id}!")
    return user_id
  else:
    print(
        "[-] Invalid User ID or Password. (Note: User ID is case-insensitive)."
    )
    return None
```

## Searching For Specific Skills 

```python

# 5. Search for students who have a specific skill
def search_skill(current_user, data):
  print("\n==============================")
  print("   SKILL SEARCH PAGE          ")
  print("==============================")

  target_skill = input("Enter the skill you want to search for (e.g., Python, Spanish, Guitar): ").strip()

  if not target_skill:
    print("Skill cannot be empty.")
    return

  print(f"\nSearching for students who know '{target_skill}'...")
  found_match = False

  for user_id, info in data.items():
    if user_id == current_user:
      continue  # Don't show yourself

    user_skills = info.get("skills", [])
    user_skills_lower = [s.lower() for s in user_skills]

    # Check if target skill matches any skill in their list
    if target_skill.lower() in user_skills_lower:
      found_match = True
      print(
          f"-> Found Student: '{user_id}' | Their Full Skills: {user_skills}"
      )

  if not found_match:
    print(f"No other students found with the skill '{target_skill}'.")
```

## Student Dashboard 

```python
# 6. Student Dashboard
def student_dashboard(current_user):
  while True:
    data = load_data()  # Reload data in case file changed

    print(f"\n==============================")
    print(f"   STUDENT DASHBOARD ({current_user})")
    print(f"==============================")
    print("1. View your own skills")
    print("2. Search for people by a skill you want to learn")
    print("3. Log out (Return to Main Menu)")

    choice = input("Choose an option (1, 2, or 3): ").strip()

    if choice == "1":
      my_skills = data.get(current_user, {}).get("skills", [])
      print(f"\nYour current skills: {my_skills}")

    elif choice == "2":
      search_skill(current_user, data)

    elif choice == "3":
      print("Logged out successfully.")
      break
    else:
      print("Invalid choice. Please enter 1, 2, or 3.")
```

## Main Menu

```python
# 7. Main Menu
def main():
  load_data()  # Initialize file on startup

  while True:
    print("\n==============================")
    print("   MAIN MENU                  ")
    print("==============================")
    print("1. Register (New User)")
    print("2. Login (Try using 'alice123' / 'pass123')")
    print("3. Exit")

    choice = input("Choose an option (1, 2, or 3): ").strip()

    if choice == "1":
      register()
    elif choice == "2":
      logged_in_user = login()
      if logged_in_user:
        student_dashboard(logged_in_user)
    elif choice == "3":
      print("Goodbye!")
      break
    else:
      print("Invalid choice. Please enter 1, 2, or 3.")


if __name__ == "__main__":
  main()
```
